## Preliminary Questions

- **Do you expect the energy split due to tunneling to increase or decrease as the energy of the eigenstates increases?**
  Since negative energy eigenstates have smaller energy than the double-well maximum ($E < V(0) = 0$), they must tunnel through the hill to be observed in either well. In Assignment 1, we showed that in a classically forbidden region where $E < V(r)$, the wavefunction decays as $\psi(r) \propto e^{-\kappa r}$, with wave number
  $$
  \kappa = \frac{\sqrt{2\mu (V(x) - E)}}{\hbar}.
  $$
  As $E \to V(0) = 0$, both $V(x) - E$ and $\kappa$ decrease, and hence the probability of tunneling increases. Since the difference in energy between the two energy states is due to the probability of tunneling, the energy split will increase as the energy of the eigenstates increases.

- **Do you still expect the energy eigenstates to be even or odd functions?**
  Since $V(x) = V(-x)$ and the system only has bound states, all energy eigenstates are even or odd. Instead of one separate state in the left well and one in the right, for each negative-energy level there are two energy states with a wave function spread across the two wells: one even, one odd.

- **If one of the lowest energy “pair” is even and the other odd, which one do you expect to have a lower energy? Discuss.**
  We expect the even eigenstate to have lower energy. The odd state satisfies $\Psi(-x) = -\Psi(x)$, so at $x = 0$, $\Psi(0) = 0$. Hence the lowest odd state has a node at $x = 0$, the middle of the hill. The lowest even state is positive across the hill and has 0 nodes, making it the ground state.
