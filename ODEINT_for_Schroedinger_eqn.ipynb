# Quantum Harmonic Oscillator

The code in the script *harm_osc_integration.py* integrates numerically 

$\psi'' = \frac{2m}{\hbar^2}(V(x) - E)\psi$

for the potential $V(x)=\frac12 m \omega^2 x^2$.

The right-hand of the Schrödinger equation is the second element of the list returned by fct() in *harm_osc_integration.py*

The *x* and *energy* in fct() are the dimensionless variables:

$x \to \zeta = \sqrt{\frac{2m\omega}{\hbar}} x$

$E \to \epsilon = \frac{E}{\hbar \omega}$

Using dimensionless variables ensures that the numbers during the
numerical integration are neither too small nor too larger to minimise numerical errors.

## The *harm_osc_integration.py* script

Move to the directory where the script*harm_osc_integration.py*
and run it from a jupyter cell:

run harm_osc_integration.py

The script will ask you:

1. if the eigenfunction is even or odd
 
2. the maximum value of x [default is 8]

3. the number of iterations you want to run [default 16]. At each iteration bracketing energies are halved.

4. The number of integration steps [default is 100]. The integration step is $\frac{x_{\rm{max}}}{\text{number of steps}}$

5. The initial lowest bracket energy

6. The initial highest bracket energy

The energy of the energy eigenstate must be inside this energy interval. If you do not know the analytical solution for the quantum system you are integrating you will have to do some 'educated' guess work, till you find a solution.  Usually it is better to start by searching the numerical solution of the ground state.

The numerical solution is in the array *psi_low* or *psi_high*. The
column [0] has the values of $\psi$ and [1] of $\psi'$. The corresponding $x$ values are in the array *x_range*.
